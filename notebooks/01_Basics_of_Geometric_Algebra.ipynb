## Basis Vectors and the Geometric Product
In 3D Euclidean space (denoted as $Cl(3,0)$), we have three orthogonal basis vectors: $e_1, e_2, e_3$. 
The fundamental rule of the **geometric product** is that the square of any basis vector is a scalar:
$$e_i^2 = 1$$
And orthogonal vectors anticommute: $$e_i e_j = -e_j e_i$$

```r
# Define the standard basis vectors for 3D space
e1 <- basis(1)
e2 <- basis(2)
e3 <- basis(3)

# The geometric product is represented by standard multiplication '*'
print("e1 * e1 is:")
print(e1 * e1)

print("e1 * e2 is:")
print(e1 * e2)
```

## The Wedge Product and Bivectors
When we wedge two vectors together ($a \wedge b$), we create a **bivector**. 
A bivector represents an oriented area, much like a vector represents an oriented length.
In 3D space, bivectors are the mathematical objects that represent planes and rotations.

```r
# The wedge product in R is represented by the '^' operator
a <- 2*e1 + 3*e2
b <- e1 - e2

area_bivector <- a %^% b
print("The bivector formed by a and b is:")
print(area_bivector)
```

## The Trivector (Pseudoscalar)
In 3D, if we wedge all three basis vectors together, we get a trivector ($e_1 \wedge e_2 \wedge e_3$), which represents an oriented volume.
This is also known as the **pseudoscalar**, often denoted as $I$. Interestingly, its square is $-1$, making it the geometric equivalent of the imaginary unit $i$.

```r
I <- e1 %^% e2 %^% e3
print("The pseudoscalar I is:")
print(I)

print("I * I is:")
print(I * I)
```
