# Part 3: Optoelectronics Applications

In Optoelectronics, we constantly deal with the manipulation of light (electromagnetic waves) using optical elements like polarizers and waveplates. 
Geometric Algebra provides a seamless way to model these transformations using **Rotors**.

```r
library(clifford)
e1 <- basis(1); e2 <- basis(2); e3 <- basis(3)
I <- e1 %^% e2 %^% e3
```

## Modeling Polarization with Rotors
Instead of using complex matrices (Jones Calculus), we can represent the polarization state of light as a vector in the $e_1-e_2$ plane. 
An optical element (like a half-wave plate) that rotates the polarization by an angle $\theta$ is represented by a **Rotor** $R$:
$$R = \cos(\theta/2) - I_{12}\sin(\theta/2)$$
Where $I_{12} = e_1 \wedge e_2$ is the bivector plane of rotation.

```r
# Define the polarization plane (e1-e2)
I12 <- e1 %^% e2

# Initial horizontally polarized light
E_in <- e1

# We want to rotate the polarization by 90 degrees (pi/2)
theta <- pi / 2

# The Rotor
R <- cos(theta/2) - I12 * sin(theta/2)
R_rev <- cos(theta/2) + I12 * sin(theta/2) # Reverse rotor

# Apply the rotation: E_out = R * E_in * R_rev
E_out <- R * E_in * R_rev

print("Input Polarization:")
print(E_in)
print("Output Polarization (should be vertical e2):")
print(E_out)
```

## Power Flow: The Poynting Vector
In Optoelectronics, measuring the intensity and direction of light propagation is critical. 
The Poynting vector $\mathbf{S}$ describes the directional energy flux.
In traditional vector calculus: $\mathbf{S} = \frac{1}{\mu_0} (\mathbf{E} \times \mathbf{B})$.
In Geometric Algebra, the cross product is replaced by taking the dual of the wedge product: $\mathbf{E} \times \mathbf{B} = -I(\mathbf{E} \wedge \mathbf{B})$.

```r
mu_0 <- 1.256e-6 # Vacuum permeability

# Electric field in x, Magnetic field in y
E_vec <- 10 * e1
B_vec <- 2 * e2

# Cross product via Geometric Algebra: -I * (E %^% B)
cross_product <- -I * (E_vec %^% B_vec)

S <- (1/mu_0) * cross_product
print("The Poynting Vector S (propagation in z-direction e3) is:")
print(S)
```
